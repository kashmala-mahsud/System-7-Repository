```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print("Original array:", arr)

reshaped_arr = np.array([[1, 2, 3], [4, 5, 6]])
print(reshaped_arr)

# Indexing and slicing
print("Element at index 2:", arr[2])
print("Slice from index 1 to 4:", arr[1:4])
print(arr[4:])
print(arr[-3:-2])
print(arr[1:5:2])
print(reshaped_arr[1, 1:2])
```

```python
# Q3. Mathematical Operations

a = np.array([10, 20, 30, 40, 50])
b = np.array([2, 4, 5, 8, 10])
```

```python
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Scalar Multiplication:", a * 2)
print("Minimum value in array a:", np.min(a))
print("Maximum value in array b:", np.max(b))
```
